# Literature Review

Before starting the EDA, I went through some research papers and articles to understand what factors can affect a student's academic performance. I found that student performance is not usually dependent on just one thing. It can be influenced by academic habits, attendance, previous performance, lifestyle, and even the student's background.

One of the main factors discussed in the research is **study habits**. Students who spend more time studying and follow a regular study routine may perform better academically. This made study hours and study methods important variables to look at in my dataset.

**Previous academic performance** is another factor that is commonly considered. A student's previous score can give some idea about their existing academic level and may have a relationship with their future results. Because of this, I included previous score while analyzing the factors related to final score.

**Attendance** is also an important factor to consider. Regular attendance gives students more opportunities to attend classes, understand concepts, and participate in classroom activities. However, the relationship between attendance and performance may not be the same for every student, so it is important to analyze the actual data instead of making assumptions.

I also found research related to **sleep and academic performance**. Sleep can affect concentration, energy, and the ability to focus while studying. Some studies have found a connection between sleep quality and academic performance, which made sleep hours another useful variable to explore in this project.

Apart from academic factors, **stress and lifestyle** can also play a role in how students perform. Students may have different stress levels, study methods, and participation in extra activities. These factors may not directly determine a student's score, but they can provide useful information when trying to understand the overall pattern in the data.

The research also discusses **socioeconomic factors**, such as family income and parents' education. These factors can give some background about a student's environment and access to educational resources. However, they should not be treated as direct causes of academic performance because student outcomes are influenced by many factors.

Overall, the literature review helped me understand that student performance is a combination of different academic, personal, behavioral, and background-related factors. This gave me a better idea of which variables to focus on during my EDA, especially study hours, attendance, previous score, assignments, sleep, stress, and other student characteristics.

It also helped me understand that finding a relationship between two variables does not necessarily mean that one variable causes the other. Therefore, in this project, I will focus on identifying patterns and relationships in the dataset rather than making causal conclusions.

# Outlier Analysis

```python
numeric_features = [
    "age", "study_hours", "attendance", "previous_score",
    "assignments_completed", "sleep_hours", "class_participation",
    "stress_level", "final_score"
]

outlier_summary = []

for column in numeric_features:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outlier_count = ((df[column] < lower_bound) | (df[column] > upper_bound)).sum()

    outlier_summary.append([
        column, Q1, Q3, lower_bound, upper_bound, outlier_count
    ])

outlier_df = pd.DataFrame(
    outlier_summary,
    columns=[
        "Column", "Q1", "Q3", "Lower Bound",
        "Upper Bound", "Outlier Count"
    ]
)

outlier_df
```

```python
plt.figure(figsize=(12, 6))

sns.boxplot(data=df[numeric_features])

plt.xticks(rotation=45)
plt.title("Outlier Analysis of Numerical Variables")
plt.xlabel("Numerical Variables")
plt.ylabel("Values")

plt.show()
```

#### Insights for outlier analysis

- The boxplot shows that most observations stay within the normal range, with a few values extending beyond the main distribution.
- *study_hours* contains some high values  , which may represents students who spend more time studying .
- *sleep_hours* also contains a few unusual observations on both lower and higher side.
